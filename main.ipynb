importing all necessary modules

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

loading scrapped data

```python
df = pd.read_csv('largest_companies.csv')
```

```python
df.head()
```

```python
df.info()
```

data cleaning

```python
df.isna().sum()
```

changing the datatypes of columns

```python

df['Revenue (USD millions)'] = (
    df['Revenue (USD millions)']
    .str.replace(',', '')
    .astype(float)
)

df['Employees'] = (
    df['Employees']
    .str.replace(',', '')
    .astype(int)
)
```

```python
df.info()
```

```python
df.head()
```

data visualization

1 ploting a bargraph on the shares of companies

```python
top20 = df.head(10)

# fig, ax = plt.subplots(figsize=(10, 6))

top20.plot(
    x='Name',
    y='Revenue (USD millions)',
    kind='bar',
    # ax=ax
)

plt.xticks(rotation=45)
plt.xlabel('Name')
plt.ylabel('Revenue (USD millions)')
plt.title('Top 20 Companies by Revenue')
plt.tight_layout()
plt.show()
```

2 ploting top companies by employees

```python
top20 = df.head(10)

# fig, ax = plt.subplots(figsize=(10, 6))

top20.plot(
    x='Name',
    y='Employees',
    kind='bar',
    # ax=ax
)

plt.xticks(rotation=45)
plt.xlabel('Name')
plt.ylabel('Employees')
plt.title('Top 20 Companies by Employee Count')
plt.tight_layout()
plt.show()
```
